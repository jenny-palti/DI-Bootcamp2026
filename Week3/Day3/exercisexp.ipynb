## 🌟 Exercise 3: Understanding Distributions
A normal distribution with mean 50 and standard deviation 10.

```python
from scipy.stats import norm

x = np.linspace(10, 90, 500)
y = norm.pdf(x, loc=50, scale=10)

plt.plot(x, y)
plt.title("Normal distribution (mean = 50, std = 10)")
plt.xlabel("Value")
plt.ylabel("Probability density")
plt.show()
```

## 🌟 Exercise 4: T-Test Application

```python
np.random.seed(42)   # so the random numbers are the same every time you run it
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = stats.ttest_ind(data1, data2)
print("T-statistic:", round(t_stat, 3))
print("P-value:", p_value)

if p_value < 0.05:
    print("The two groups have significantly different means.")
else:
    print("No significant difference between the two groups.")
```

## 🌟 Exercise 5: Linear Regression Analysis

```python
house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

result = stats.linregress(house_sizes, house_prices)
print("Slope:", round(result.slope, 2))
print("Intercept:", round(result.intercept, 2))

predicted_price = result.slope * 90 + result.intercept
print("Predicted price for 90 m²:", round(predicted_price, 2))
```

**Interpretation of the slope:** the slope is about **1,810**, meaning every extra square meter adds roughly **1,810 currency units** to the price of a house, according to this data. The intercept (~64,000) is the model's starting point; it isn't meaningful on its own, because a 0 m² house doesn't exist.

## 🌟 Exercise 6: Understanding ANOVA

```python
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print("F-value:", round(f_value, 3))
print("P-value:", p_value)
```

**Answers:**
- The P-value is far below 0.05, so the fertilizers **do have significantly different effects** on plant growth (fertilizer 2 grows the most, fertilizer 3 the least).
- **If the P-value were greater than 0.05**, we would *not* reject the null hypothesis: we couldn't say the fertilizers differ, and any differences between the groups could just be random variation.

## 🌟 Exercise 8: Correlation Coefficients (Optional)

```python
data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})

pearson_r, pearson_p = stats.pearsonr(data["age"], data["income"])
spearman_r, spearman_p = stats.spearmanr(data["age"], data["income"])

print("Pearson correlation:", round(pearson_r, 4))
print("Spearman correlation:", round(spearman_r, 4))
```

Both are close to **1**: income rises almost perfectly with age in this small example. Spearman is exactly 1 because the *ranks* match perfectly (older always means higher income); Pearson is slightly below 1 because the relationship isn't a perfectly straight line.
