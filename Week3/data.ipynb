```python
dayofweek = [1,2,3,4,5,6,7]
temperature = [28, 29, 33, 27, 30, 29, 31]
plt.plot (dayofweek, temperature, color = 'orange')
plt.title ('Temperatures this week')
plt.xlabel ('Days')
plt.ylabel ('Temperature')
plt.show()
```
