# Daily Challenge – Week 4, Day 1: NumPy, Pandas and Matplotlib Integration
Scenario: analyze (synthetic) average monthly temperatures for 10 cities.

## 1. Data Preparation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)   # fixed seed, so the "random" data is the same every run

cities = ["Tel Aviv", "London", "New York", "Tokyo", "Sydney",
          "Moscow", "Cairo", "Rio de Janeiro", "Reykjavik", "Mumbai"]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# 10 cities x 12 months, uniformly between -5 and 35 °C
temperatures = np.random.uniform(low=-5, high=35, size=(10, 12)).round(1)

df = pd.DataFrame(temperatures, index=cities, columns=months)
df
```

## 2. Data Analysis

```python
df["Annual Avg"] = df[months].mean(axis=1)   # axis=1 → average across the months of each row (city)
df["Annual Avg"].sort_values(ascending=False).round(2)
```

```python
hottest = df["Annual Avg"].idxmax()
coldest = df["Annual Avg"].idxmin()
print(f"Highest annual average: {hottest} ({df.loc[hottest, 'Annual Avg']:.2f} °C)")
print(f"Lowest annual average:  {coldest} ({df.loc[coldest, 'Annual Avg']:.2f} °C)")
```

```python
# Extra: which month is warmest on average across all cities, and how much does each city vary?
monthly_avg = df[months].mean(axis=0)
print("Warmest month (all cities):", monthly_avg.idxmax(), round(monthly_avg.max(), 2))
print("Coldest month (all cities):", monthly_avg.idxmin(), round(monthly_avg.min(), 2))

df["Range"] = df[months].max(axis=1) - df[months].min(axis=1)
df["Range"].sort_values(ascending=False).round(1)
```

## 3. Data Visualization

```python
plt.figure(figsize=(12, 6))
for city in cities:
    plt.plot(months, df.loc[city, months], marker="o", label=city)
plt.title("Monthly average temperature by city")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
df["Annual Avg"].sort_values().plot(kind="barh", figsize=(8, 5), color="tomato")
plt.title("Annual average temperature by city")
plt.xlabel("Temperature (°C)")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.imshow(df[months], cmap="coolwarm", aspect="auto")
plt.colorbar(label="°C")
plt.xticks(range(12), months)
plt.yticks(range(10), cities)
plt.title("Heatmap of monthly temperatures")
plt.show()
```

## 4. Summary Report
- **Highest annual average:** Rio de Janeiro (**17.61 °C**). **Lowest:** London (**9.58 °C**). The gap between the warmest and coldest city is about 8 °C.
- **Warmest month across all cities:** August (~19.7 °C). **Coldest:** July (~8.2 °C), which is unrealistic and shows the data is random.
- **Variation within each city is huge:** most cities swing **30–38 °C** between their hottest and coldest month (Tel Aviv the most, 38 °C), and the line chart jumps up and down from month to month.
- **Important caveat:** the temperatures were generated with `np.random.uniform`, so every month and city is independent noise between −5 and 35 °C. That's why there are **no real seasonal patterns** (no summer peak, no northern/southern hemisphere difference) and why "Reykjavik" can be warmer than "Tokyo". The city names are labels only. With real weather data, the same code would show smooth seasonal curves and real differences between climates.
- **How each library was used:** NumPy generated the data and did the averages; Pandas organized it into a table (cities × months) and found the extremes with `idxmax()` / `idxmin()`; Matplotlib drew the line chart, bar chart and heatmap.
