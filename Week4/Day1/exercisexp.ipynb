## 🌟 Exercise 1: Array Creation and Manipulation
A 1D array with the numbers 0 to 9.

```python
arr = np.arange(10)
arr
```

## 🌟 Exercise 2: Type Conversion and Array Operations

```python
floats = np.array([3.14, 2.17, 0, 1, 2])
ints = floats.astype(int)   # astype cuts off the decimals
ints
```

## 🌟 Exercise 3: Working with Multi-Dimensional Arrays
A 3×3 array with the values 1 to 9.

```python
matrix = np.arange(1, 10).reshape(3, 3)
matrix
```

## 🌟 Exercise 4: Creating a Multi-Dimensional Array with Random Numbers
Shape (4, 5), random numbers between 0 and 1 (rounded to 2 decimals to match the example).

```python
random_array = np.random.rand(4, 5).round(2)
random_array
```

## 🌟 Exercise 5: Indexing Arrays
Select the **second** row (index 1, because counting starts at 0).

```python
array = np.array([[21, 22, 23, 22, 22], [20, 21, 22, 23, 24], [21, 22, 23, 22, 22]])
array[1]
```

## 🌟 Exercise 6: Reversing Elements

```python
arr = np.arange(10)
arr[::-1]   # [start:stop:step] with step -1 walks backwards
```

## 🌟 Exercise 7: Identity Matrix

```python
np.eye(4)
```

## 🌟 Exercise 8: Simple Aggregate Functions

```python
arr = np.arange(10)
print(f"Sum: {arr.sum()}, Average: {arr.mean()}")
```

## 🌟 Exercise 9: Create an Array and Change its Structure

```python
arr = np.arange(1, 21)
arr.reshape(4, 5)
```

## 🌟 Exercise 10: Conditional Selection of Values
Extract the odd numbers. `arr % 2 == 1` makes a True/False mask, and `arr[mask]` keeps only the True ones.

```python
arr = np.arange(10)
arr[arr % 2 == 1]
```
