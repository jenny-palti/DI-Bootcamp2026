## 🌟 Exercise 1: Matrix Operations
The **determinant** tells us whether a matrix can be inverted (it can only if the determinant is not 0).
The **inverse** A⁻¹ is the matrix that gives the identity matrix when multiplied by A.

```python
A = np.array([[2, 1, 1],
              [1, 3, 2],
              [1, 0, 0]])

det = np.linalg.det(A)
print("Determinant:", round(det, 2))

A_inv = np.linalg.inv(A)
print("Inverse:\n", A_inv.round(2))

# Check: A × A⁻¹ should be the identity matrix
print("A @ A_inv:\n", (A @ A_inv).round(2))
```

## 🌟 Exercise 5: Image Representation
A **grayscale image** is just a 2D NumPy array: each element is one pixel, and its value is the brightness (0 = black, 255 = white).
A **color image** is a 3D array of shape (height, width, 3): one layer each for red, green and blue.

```python
image = np.zeros((5, 5), dtype=np.uint8)   # 5x5 black image
image[1:4, 1:4] = 128                      # grey square in the middle
image[2, 2] = 255                          # white pixel in the center
print(image)

plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.title("5x5 grayscale image")
plt.axis("off")
plt.show()
```

## 🌟 Exercise 6: Basic Hypothesis Testing
- **H0:** the training program has **no effect**: the average productivity after = before.
- **H1:** productivity is **higher** after the training.

The same employees are measured twice, so we use a **paired t-test** on the differences.

```python
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

differences = productivity_after - productivity_before
print("Average before:", round(productivity_before.mean(), 2))
print("Average after:", round(productivity_after.mean(), 2))
print("Average improvement:", round(differences.mean(), 2))

# t-statistic by hand with NumPy: mean difference / (std / sqrt(n))
n = len(differences)
t_manual = differences.mean() / (differences.std(ddof=1) / np.sqrt(n))
print("t-statistic (NumPy):", round(t_manual, 2))

# Same test with SciPy, to get the p-value
t_stat, p_value = stats.ttest_rel(productivity_after, productivity_before, alternative="greater")
print("t-statistic (SciPy):", round(t_stat, 2), "| p-value:", p_value)

if p_value < 0.05:
    print("Reject H0: the training program significantly increased productivity.")
else:
    print("Cannot reject H0: no significant effect.")
```

## 🌟 Exercise 8: Time Series Data Manipulation
One value per day for 2023, sliced by quarter.

```python
dates = pd.date_range("2023-01-01", "2023-12-31", freq="D")
series = pd.Series(np.random.randn(len(dates)).cumsum(), index=dates)

q1 = series["2023-01":"2023-03"]
q2 = series["2023-04":"2023-06"]
q3 = series["2023-07":"2023-09"]
q4 = series["2023-10":"2023-12"]

for name, part in [("Jan–Mar", q1), ("Apr–Jun", q2), ("Jul–Sep", q3), ("Oct–Dec", q4)]:
    print(f"{name}: {len(part)} days, from {part.index[0].date()} to {part.index[-1].date()}")
```
