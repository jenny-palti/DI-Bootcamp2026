## 1. What is Machine Learning, and why does it matter for data analysts?
**Machine Learning (ML)** is a branch of artificial intelligence in which a computer **learns patterns from data** instead of following rules that a programmer wrote by hand. We give an algorithm many examples; it finds the relationships in them and builds a **model** that can then make predictions or decisions about new data it has never seen.

For example, instead of writing rules like "if income < X and debt > Y, the loan is risky", we show a model thousands of past loans together with whether they were repaid, and it learns on its own which combinations of factors signal risk.

**Why it matters for data analysts:**
- **From describing to predicting:** classic analysis explains what *happened* (reports, dashboards). ML lets analysts estimate what *will* happen – which customers will leave, which transactions are fraud, how much will sell next month.
- **Scale and complexity:** ML can find patterns across hundreds of variables and millions of rows, which no person could spot by looking at charts.
- **Automation:** once trained, a model can score new data instantly and repeatedly (e.g. every new loan application).
- **Better decisions:** analysts can turn data into concrete, measurable recommendations, and test how good those predictions are.
- **Analysts are essential to ML:** ML depends on clean, well-understood data. Cleaning, exploring, choosing features and interpreting results are core analyst skills, and they decide whether a model is useful or misleading.

## 2. Applications of ML across industries
1. **Healthcare – early disease detection.** Models trained on medical images (X-rays, retina scans, mammograms) flag likely signs of disease so that doctors can review the most urgent cases first. In mental health, models can analyze text or app usage to help identify people who may need support sooner.
2. **Finance – fraud detection and credit scoring.** Banks score every card transaction in real time and block those that look unusual compared to a customer's normal behaviour. ML is also used to estimate the risk that a borrower will not repay a loan.
3. **Retail & e-commerce – recommendations and demand forecasting.** Platforms like Amazon or Netflix recommend products and shows based on what similar users liked. Retailers forecast demand per product and store, so they order the right amount of stock.
4. **Transportation – navigation and self-driving.** Map apps predict travel times from live traffic data, and self-driving systems learn to recognise pedestrians, signs and other cars.

## 3. Types of Machine Learning

| Type | How it learns | Typical tasks | Example scenario |
|---|---|---|---|
| **Supervised learning** | From **labeled** data: every example comes with the correct answer (the *target*). The model learns to map inputs → answer. | Classification (categories), Regression (numbers) | **Spam filter:** trained on emails already labeled "spam" / "not spam", it predicts the label of new emails. Also: predicting house prices from size and location. |
| **Unsupervised learning** | From **unlabeled** data: there is no correct answer. The model looks for structure on its own. | Clustering, dimensionality reduction (e.g. PCA), anomaly detection | **Customer segmentation:** a shop groups customers by purchase behaviour (e.g. "bargain hunters", "loyal big spenders") without telling the algorithm what the groups are. |
| **Reinforcement learning** | By **trial and error** in an environment: an *agent* takes actions and receives **rewards** or penalties, and learns the strategy that maximises total reward. | Games, robotics, control, recommendation strategies | **Game-playing AI:** AlphaGo learned to play Go by playing millions of games and being rewarded for winning. Also: a robot learning to walk or find its way through a maze. |

**In one sentence each:**
- *Supervised* = learn from examples **with** answers.
- *Unsupervised* = find patterns **without** answers.
- *Reinforcement* = learn from **rewards** for actions.

## 4. The process of developing an ML model
Before these three stages there is always **problem definition** (what exactly are we predicting?) and **data collection & cleaning**.

### Stage 1 – Feature Selection
**Features** are the input variables the model learns from (e.g. income, age, credit history). Feature selection means choosing the ones that actually help prediction.
- Remove features that are irrelevant, duplicated, or that "leak" the answer (information you wouldn't have at prediction time).
- Use domain knowledge, correlations, statistical tests, or model-based importance scores to choose.
- Often create new features (*feature engineering*), e.g. "loan amount ÷ income".
- **Why it matters:** fewer, better features → simpler, faster models that generalise better and are easier to explain.

### Stage 2 – Model Selection
Choose the algorithm that fits the problem and the data.
- **Type of problem:** classification → e.g. logistic regression, decision tree, random forest; regression → linear regression, gradient boosting; clustering → K-means.
- **Trade-offs:** simple models (logistic regression) are easy to explain; complex ones (random forest, neural networks) can be more accurate but are harder to interpret and need more data.
- In practice, several candidate models are trained and compared, and their settings (*hyperparameters*) are tuned.

### Stage 3 – Model Evaluation
Check how well the model performs on data it has **not** seen during training.
- **Split the data** into training and test sets (e.g. 80/20), and use **cross-validation** for more reliable estimates.
- **Choose metrics that fit the problem:** accuracy, precision, recall, F1-score, ROC-AUC for classification; MAE / RMSE / R² for regression; silhouette score for clustering.
- Watch for **overfitting** (great on training data, poor on new data) and underfitting.
- If results aren't good enough, go back: improve features, try other models, or collect more data. The process is a **loop**, not a straight line.

### Flowchart of the ML process

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

steps = ["1. Define the\nproblem", "2. Collect &\nclean data", "3. Feature\nselection",
         "4. Model\nselection", "5. Train\nthe model", "6. Model\nevaluation", "7. Deploy &\nmonitor"]
colors = ["#d9d9d9", "#d9d9d9", "#9ecae1", "#9ecae1", "#d9d9d9", "#9ecae1", "#d9d9d9"]

fig, ax = plt.subplots(figsize=(15, 3.6))
ax.set_xlim(0, 15)
ax.set_ylim(-0.2, 3.4)
ax.axis("off")

for i, (step, color) in enumerate(zip(steps, colors)):
    x = 0.3 + i * 2.1
    ax.add_patch(FancyBboxPatch((x, 1.6), 1.7, 1.2, boxstyle="round,pad=0.08", facecolor=color, edgecolor="black"))
    ax.text(x + 0.85, 2.2, step, ha="center", va="center", fontsize=10)
    if i < len(steps) - 1:
        ax.annotate("", xy=(x + 2.05, 2.2), xytext=(x + 1.8, 2.2), arrowprops=dict(arrowstyle="->", lw=1.5))

# Feedback loop: from evaluation back to feature selection
ax.annotate("", xy=(5.35, 1.45), xytext=(11.65, 1.45),
            arrowprops=dict(arrowstyle="->", lw=1.5, color="firebrick", connectionstyle="arc3,rad=-0.3"))
ax.text(8.5, 0.15, "Not good enough? Improve features, try another model, tune it", ha="center",
        color="firebrick", fontsize=10)

ax.set_title("Machine learning model development process (blue = the three main stages)", fontsize=12)
plt.show()
```

## Summary
Machine learning lets computers learn from data to make predictions or find structure. It comes in three main types – supervised (learn from labeled answers), unsupervised (discover patterns on their own) and reinforcement (learn from rewards). Building a model follows a repeatable loop: choose good **features**, pick a suitable **model**, and **evaluate** it honestly on unseen data – then iterate. For data analysts, ML extends their work from explaining the past to predicting the future, and their skills in understanding and preparing data are what make ML models trustworthy.
